# 隨機分類器

```python
import os
import random
import shutil
```

```python
PATH_HC = 'C:/Users/kitt9/OneDrive - National ChengChi University/eeg-rag-llm-diag/origin_data_set/HC/'
PATH_MCI = 'C:/Users/kitt9/OneDrive - National ChengChi University/eeg-rag-llm-diag/origin_data_set/MCI/'
PATH_AD = 'C:/Users/kitt9/OneDrive - National ChengChi University/eeg-rag-llm-diag/origin_data_set/AD/'
```

```python
TARGET_ROOT = './sample_test'  # 目的地的根目錄名稱

source_paths = {
    'HC': PATH_HC,
    'MCI': PATH_MCI,
    'AD': PATH_AD
}

# 固定隨機種子，確保如果你之後重新執行，檔案分流的結果會一模一樣（便於實驗追蹤）
random.seed(42)

print("=" * 60)
# 自動建立 sample 與 test 的基礎階層架構
for part in ['sample', 'test']:
    for group in source_paths.keys():
        os.makedirs(os.path.join(TARGET_ROOT, part, group), exist_ok=True)
print(f"🎉 已成功初始化 '{TARGET_ROOT}' 階層資料夾結構。")
print("=" * 60)

# ==============================================================================
# 核心邏輯：各組獨立盲抽 80% 檔案至 sample，剩下 20% 至 test
# ==============================================================================
for group, src_dir in source_paths.items():
    if not os.path.exists(src_dir):
        print(f"❌ 錯誤：找不到來源資料夾 {src_dir}，跳過此組別。")
        continue
        
    # 撈出該資料夾下所有 .edf 或 .EDF 的檔案
    all_files = [f for f in os.listdir(src_dir) if f.lower().endswith('.edf')]
    total_count = len(all_files)
    
    if total_count == 0:
        print(f"⚠️ 提示：組別 {group} 資料夾中沒有找到任何 EDF 檔案。")
        continue
        
    # 將檔案清單順序隨機打亂
    random.shuffle(all_files)
    
    # 計算 80% 的切分點數量
    sample_size = int(total_count * 0.8)
    
    sample_files = all_files[:sample_size]
    test_files = all_files[sample_size:]
    
    print(f"組別 {group:7s}: 總檔案數 = {total_count:3d} 檔 -> 隨機分配 sample = {len(sample_files):3d} 檔 | test = {len(test_files):3d} 檔")
    
    # 執行實體檔案複製 (放進 sample/)
    for filename in sample_files:
        src_path = os.path.join(src_dir, filename)
        dest_path = os.path.join(TARGET_ROOT, 'sample', group, filename)
        shutil.copy2(src_path, dest_path)
        
    # 執行實體檔案複製 (放進 test/)
    for filename in test_files:
        src_path = os.path.join(src_dir, filename)
        dest_path = os.path.join(TARGET_ROOT, 'test', group, filename)
        shutil.copy2(src_path, dest_path)

print("=" * 60)
print(f"🎉 實體檔案隨機分流搬移完成！請至 '{TARGET_ROOT}' 資料夾查看結果。")
print("=" * 60)
```
